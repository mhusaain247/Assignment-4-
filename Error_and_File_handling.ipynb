### Q4

### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
file = open("numbers.txt", "r")
total = 0
for line in file:
    total = total + int(line)
file.close()
print(total)
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# data.txt file banayi taake read ho sake
with open("data.txt", "w") as file:
    file.write("This is data file.")

with open("data.txt", "r") as file:
    content = file.read()
    print(content)

# Faida: with statement file ko kaam khatam hone par khud band (close) kar deta hai,
# chahe beech mein error aa jaye. Isliye file.close() likhne ki zaroorat nahi parti.
```

### Q8 (Medium)

### Q9

### Q10 
`os` module ka use karke check karein ki kya `test.txt` file exist karti hai ya nahi. Agar karti hai toh usay delete kar dein.

### Q12 (Easy)
User se number input lein. Agar user text enter kar de toh `ValueError` handle karke custom error message print karein.

```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("Error: Please enter a valid number, not text")
```

### Q13 (Easy)

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

### Q15 
Ek list `my_list = [10, 20, 30]` banayein. Invalid index (e.g., index 5) access karein aur `IndexError` catch karein.

```python
my_list = [10, 20, 30]

try:
    print(my_list[5])
except IndexError:
    print("Error: Index out of range")
```

### Q16 
Ek dictionary `student = {"name": "Ali"}` banayein. Non-existent key (`"age"`) access karein aur `KeyError` handle karein.

```python
student = {"name": "Ali"}

try:
    print(student["age"])
except KeyError:
    print("Error: Key 'age' not found")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
except ValueError:
    print("Error: Please enter a valid number")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
else:
    print("No error occurred. Result:", result)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print("Age is valid")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file = None

try:
    file = open("data.txt", "r")
    print(file.read())
except FileNotFoundError:
    print("Error: File not found")
finally:
    if file != None:
        file.close()
        print("File closed")
```
